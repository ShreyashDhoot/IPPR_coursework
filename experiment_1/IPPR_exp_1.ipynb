*  ADDING 2 IMAGES

```python
# load and convert image to RGB and resize for addition
image=cv2.imread("D:/ippr/experiment_1/image.jpg")
image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
image=cv2.resize(image,(500,500))
# load and convert image to RGB and resize for addition
image2=cv2.imread("D:/ippr/experiment_1/image2.jpg")
image2=cv2.cvtColor(image2,cv2.COLOR_BGR2RGB)
image2=cv2.resize(image2,(500,500))
# add 2 images
added_image= cv2.add(image,image2,)

#plot images
plt.figure(figsize=(12,8))
plt.subplot(1,3,1)
plt.imshow(image)
plt.title("original image")

plt.subplot(1,3,2)
plt.imshow(image2)
plt.title("original image 2")

plt.subplot(1,3,3)
plt.imshow(added_image)
plt.title("added image")
```

*  Translating an image

```python
height,width=image.shape[:2]
height_shift,width_shift=height/3,width/3

#Translation matrix
T=np.float32([[1,0,width_shift],
              [0,1,height_shift]])

translated_image=cv2.warpAffine(image, # <=image for translation
                                T, # <= translation matrix
                                (width,height))

plt.figure(figsize=(12,9))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title("original image")

plt.subplot(1,2,2)
plt.imshow(translated_image)
plt.title("translated image")
```

*  Scaling an Image

```python
scaling_down=0.0623
scaling_up=0.8

resize_down=cv2.resize(image,None,fx=scaling_down,fy=scaling_down,interpolation=cv2.INTER_AREA)
resize_up=cv2.resize(image,None,fx=scaling_up,fy=scaling_up,interpolation=cv2.INTER_AREA)

plt.figure(figsize=(12,9))
plt.subplot(1,3,1)
plt.imshow(image)
plt.title("original image")

plt.subplot(1,3,2)
plt.imshow(resize_up)
plt.title("scaled up image")

plt.subplot(1,3,3)
plt.imshow(resize_down)
plt.title("scaled down image")
```

```python
size = 400
block = size // 2

chess_2x2 = np.zeros((size, size), dtype=np.uint8)

chess_2x2[0:block, 0:block] = 255
chess_2x2[block:size, block:size] = 255

plt.imshow(chess_2x2, cmap='gray')
plt.title("2x2 Chessboard")
plt.axis('off')
plt.show()
```

```python
size = 400
diag_img = np.zeros((size, size), dtype=np.uint8)

for i in range(size):
    for j in range(size):
        if i > j:
            diag_img[i, j] = 255

plt.imshow(diag_img, cmap='gray')
plt.title("Diagonal Separation Image")
plt.axis('off')
plt.show()
```

```python
size = 400
squares = 8
block = size // squares

full_chess = np.zeros((size, size), dtype=np.uint8)

for i in range(squares):
    for j in range(squares):
        if (i + j) % 2 == 0:
            full_chess[i*block:(i+1)*block, j*block:(j+1)*block] = 255

plt.imshow(full_chess, cmap='gray')
plt.title("Full Chessboard (8x8)")
plt.axis('off')
plt.show()
```

```python
# defining break points 
r1,r2=70,180
s1,s2=30,250
```

```python
img_scaled = np.zeros_like(img_1_greyscale, dtype=np.float32)

# Region 1: 0 <= r <= r1
mask1 = (img_1_greyscale >= 0) & (img_1_greyscale <= r1)
img_scaled[mask1] = (s1 / r1) * img_1_greyscale[mask1]

# Region 2: r1 < r <= r2
mask2 = (img_1_greyscale > r1) & (img_1_greyscale<= r2)
img_scaled[mask2] = ((s2 - s1) / (r2 - r1)) * (img_1_greyscale[mask2] - r1) + s1

# Region 3: r2 < r <= 255
mask3 = img_1_greyscale > r2
img_scaled[mask3] = ((255 - s2) / (255 - r2)) * (img_1_greyscale[mask3] - r2) + s2

# Convert back to uint8
img_scaled = np.clip(img_scaled, 0, 255).astype(np.uint8)
```

```python
plt.subplot(1,2,1)
plt.imshow(img_scaled,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img_1_greyscale,cmap='gray')
```

```python
histogram=np.histogram(img_1_greyscale,bins=256,range=[0,255])
plt.plot(histogram[1][:-1], histogram[0])
histogram_2=np.histogram(img_scaled,bins=256,range=[0,255])
plt.plot(histogram_2[1][:-1], histogram_2[0])
```

```python
L,U=100,180

sliced = np.zeros_like(img_1_greyscale)
sliced[(img_1_greyscale >= L) & (img_1_greyscale <= U)] = 255

# Display
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(img_1_greyscale, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Gray Level Slicing (No Background)")
plt.imshow(sliced, cmap='gray')
plt.axis('off')
plt.show()
```
